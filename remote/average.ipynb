```python
from pyspark.sql import SparkSession, functions as F

spark = SparkSession.builder.appName('Average').getOrCreate()
```

```python
movies_rating = spark.read.csv('/Volumes/workspace/default/volume/movie_ratings.csv', schema="movie STRING, rating INT")

display(movies_rating)
```

```python
avg_rating = movies_rating.groupBy('movie')\
                          .agg(
                              F.round(F.avg('rating')).alias('avg_rating')
                          )
avg_rating.orderBy(F.col('movie').asc()).show()
```
