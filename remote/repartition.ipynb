```python
from pyspark.sql import functions as F, SparkSession

spark = SparkSession.builder.appName("Repartition").getOrCreate()
text_df = spark.read.text('/Volumes/workspace/default/volume/sample.txt')
```

```python
word_count_df = text_df\
    .select(F.explode(F.split(F.col('value'), ' ')).alias('word'))\
    .groupBy('word').agg(F.count('*').alias('count'))

display(word_count_df)
```

```python
word_count_df.repartition(5).write.parquet('/Volumes/workspace/default/volume/output/FivePartitionOutput')
```

```python
read_parquet_result = spark.read.parquet('/Volumes/workspace/default/volume/output/FivePartitionOutput')

display(read_parquet_result)
```
