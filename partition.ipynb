```python
from pyspark import SparkConf, SparkContext

conf = SparkConf().setAppName('Partition')
sc = SparkContext.getOrCreate(conf)
```

```python
rdd = sc.textFile('quiz3.txt')
rdd.getNumPartitions()
```

```python
rdd = rdd.repartition(5)
rdd.getNumPartitions()
```

```python
rdd2 = rdd.flatMap(lambda x: x.split(' '))
rdd3 = rdd2.map(lambda x: (x, 1))
rdd3.saveAsTextFile('output/FivePartitionOutput')
```

```python
rdd3 = rdd3.coalesce(2)
rdd3.saveAsTextFile('output/TwoPartitionOutput')
```

```python
rdd = sc.textFile('output/FivePartitionOutput')
rdd.collect()
```
