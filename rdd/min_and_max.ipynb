```python
from pyspark import SparkConf, SparkContext

conf = SparkConf().setAppName('Min and Max')
sc = SparkContext.getOrCreate(conf)
```

```python
rdd = sc.textFile('movie_ratings.csv')
rdd.collect()
```

```python
rdd2 = rdd.map(lambda x: (x.split(',')[0], x.split(',')[1]))\
          .reduceByKey(lambda x, y: min(x, y))

rdd2.collect()
```

```python
rdd3 = rdd.map(lambda x: (x.split(',')[0], x.split(',')[1]))\
          .reduceByKey(lambda x, y: max(x, y))

rdd3.collect()
```

## Quiz

```python
quiz_rdd = sc.textFile('average_quiz_sample.csv')

quiz_rdd = quiz_rdd.map(lambda x: (x.split(',')[1], x.split(',')[2]))
quiz_rdd.collect()
```

```python
min_rdd = quiz_rdd.reduceByKey(lambda x, y: min(x, y))
max_rdd = quiz_rdd.reduceByKey(lambda x, y: max(x, y))

print(min_rdd.collect())
print(max_rdd.collect())
```
