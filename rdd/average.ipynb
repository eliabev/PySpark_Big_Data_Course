```python
from pyspark import SparkConf, SparkContext

conf = SparkConf().setAppName('Average')
sc = SparkContext.getOrCreate(conf)
```

```python
rdd = sc.textFile('movie_ratings.csv')
rdd.collect()
```

```python
rdd2 = rdd.map(lambda x: (x.split(',')[0], (int(x.split(',')[1]), 1)))\
          .sortBy(lambda x: (x[0], x[1]), ascending=True)
```

```python
rdd3 = rdd2.reduceByKey(lambda acc, x: (acc[0] + x[0], acc[1] + x[1])).mapValues(lambda x: round(x[0] / x[1], 3))
rdd3.collect()
```

## Quiz

```python
def get_quiz_columns(x):
    a, b, c = x.split(',')
    return (a, (float(c), 1))

quiz_rdd = sc.textFile('average_quiz_sample.csv').map(get_quiz_columns)
quiz_rdd.collect()
```

```python
reduced_rdd = quiz_rdd\
                .reduceByKey(lambda acc, x: (acc[0] + x[0], acc[1] + x[1]))\
                .mapValues(lambda x: round(x[0] / x[1], 3))
reduced_rdd.collect()
```
