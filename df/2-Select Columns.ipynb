```python
from pyspark.sql import functions as F, SparkSession

spark = SparkSession.builder.appName('Select Columns').getOrCreate()
```

```python
students_df = spark\
                .read\
                .option('header', 'true')\
                .schema('age INT, gender STRING, name STRING, course STRING, roll STRING, marks INT, email STRING')\
                .csv('/Volumes/workspace/default/volume/StudentData.csv')

students_df.show()
```

```python
students_df.select("age", 'course').show()
```

```python
students_df.select(students_df.age).show()
```

```python
students_df.select('*').show()
```

```python
students_df.select(students_df.columns[-1::-1]).show()
```

```python
students_df.groupBy('name', 'course').count().orderBy(F.col('count').desc()).show()
```
