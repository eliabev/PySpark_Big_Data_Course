```python
from pyspark.sql import functions as F, SparkSession

spark = SparkSession.builder.appName('With Columns').getOrCreate()
```

```python
students_df = spark\
                .read\
                .option('header', 'true')\
                .csv('/Volumes/workspace/default/volume/StudentData.csv')

students_df.show()
```

```python
students_df = students_df.withColumn('age', F.col('age').cast('int'))
students_df.printSchema()
```

```python
students_df = students_df.withColumn('marks', F.col('marks') + 10)
students_df.orderBy(F.col('marks').desc()).show()
```

```python
students_df = students_df.withColumn('aggregated_marks', F.col('marks').cast('int') + 10)
students_df.show()
```

```python
students_df.printSchema()
```

```python
students_df = students_df.withColumn('country', F.lit('Brazil'))
students_df.show(30)
```
