```python
from pyspark.sql import SparkSession, functions as F

spark = SparkSession.builder.appName('Min and Max').getOrCreate()
```

```python
movies_rating = spark.read.csv('/Volumes/workspace/default/volume/movie_ratings.csv', schema='movie String, rating INT')

min_max_ratings = movies_rating.groupBy('movie').agg(
    F.min('rating').alias('min_rating'),
    F.max('rating').alias('max_rating')
)

min_max_ratings.show()
```

## Quiz

```python
quiz_df = spark.read.csv('/Volumes/workspace/default/volume/average_quiz_sample.csv', schema='month STRING, city STRING, rating FLOAT')

min_max_quiz_df = quiz_df.groupBy('city').agg(
    F.min('rating').alias('min_rating'),
    F.max('rating').alias('max_rating')
)

display(min_max_quiz_df)
```
