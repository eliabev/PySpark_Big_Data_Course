## First DataFrame actually created by the course

```python
from pyspark.sql import functions as F, SparkSession

spark = SparkSession.builder.appName('Spark DataFrame').getOrCreate()
```

```python
students_df = spark\
                .read\
                .option('header', 'true')\
                .schema('age INT, gender STRING, name STRING, course STRING, roll STRING, marks INT, email STRING')\
                .csv('/Volumes/workspace/default/volume/StudentData.csv')

students_df.show()
```

```python
students_df.printSchema()
```

```python
"""
with "options" we can even read a tsv using csv method and changing the delimiter

    options(..., delimiter='\t')

"""
students_df = spark\
                .read\
                .options(header='true', inferSchema=True)\
                .csv('/Volumes/workspace/default/volume/StudentData.csv')

students_df.printSchema()
```
