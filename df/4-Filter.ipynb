```python
from pyspark.sql import functions as F, SparkSession

spark = SparkSession.builder.appName('Filter/Where').getOrCreate()
```

```python
students_df = spark\
                .read\
                .option('header', 'true')\
                .schema('age INT, gender STRING, name STRING, course STRING, roll STRING, marks INT, email STRING')\
                .csv('/Volumes/workspace/default/volume/StudentData.csv')

students_df.show()
```

```python
students_df.filter(
    (students_df.course == 'DB') & (
        (students_df.marks > 80) | (students_df.gender == 'Female')
        )
    ).show()
```

```python
students_df\
    .filter(students_df.marks > 50)\
    .groupBy(students_df.course)\
    .agg(F.count('*').alias('passed_per_course'))\
    .orderBy(F.col('passed_per_course').desc())\
    .show()
```

```python
courses = ['DB', 'Cloud', 'DSA']

students_df\
    .filter(students_df.course.isin(courses))\
    .show()
```

```python
students_df\
    .filter(students_df.course.startswith('D'))\
    .distinct()\
    .show()
```

```python
students_df\
    .filter(students_df.name.contains('n'))\
    .distinct()\
    .show()
```

```python
students_df\
    .filter(F.upper(students_df.name).contains('SE'))\
    .distinct()\
    .show()
```

```python
students_df\
    .filter(students_df.name.ilike('%se%'))\
    .distinct()\
    .show()
```
