```python
from pyspark.sql import functions as F, SparkSession

spark = SparkSession.builder.appName('Filter/Where').getOrCreate()
```

```python
students_df = spark\
                .read\
                .option('header', 'true')\
                .schema('age INT, gender STRING, name STRING, course STRING, roll STRING, marks INT, email STRING')\
                .csv('/Volumes/workspace/default/volume/StudentData.csv')

students_df.show()
```

```python
students_df = students_df\
                .withColumn('total_marks', F.lit(120))\
                .withColumn('average_marks', F.round((F.col('marks')/F.col('total_marks')) * 100, 3))
students_df.show()
```

```python
passed_oop_df = students_df.where((students_df.course == 'OOP') & (students_df.average_marks > 80))
passed_oop_df.select('name', 'marks', 'total_marks', 'average_marks').show()
```

```python
passed_cloud_df = students_df.where((students_df.course == 'Cloud') & (students_df.average_marks > 60))
passed_cloud_df.select('name', 'marks', 'total_marks', 'average_marks').show()
```
